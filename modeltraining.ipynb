```python
import os
import numpy as np
import librosa

# Dataset path
dataset_path = "audio_speech_actors_01-24"

# Emotion mapping
emotion_map = {
    "01": "neutral",
    "02": "calm",
    "03": "happy",
    "04": "sad",
    "05": "angry",
    "06": "fearful",
    "07": "disgust",
    "08": "surprised"
}

features = []
labels = []

# Go through all actor folders
for actor in sorted(os.listdir(dataset_path)):

    actor_path = os.path.join(dataset_path, actor)

    if not os.path.isdir(actor_path):
        continue

    # Go through all audio files
    for file in os.listdir(actor_path):

        if file.endswith(".wav"):

            file_path = os.path.join(actor_path, file)

            # Extract emotion code
            emotion_code = file.split("-")[2]
            emotion = emotion_map[emotion_code]

            try:
                # Load audio
                signal, sample_rate = librosa.load(
                    file_path,
                    sr=22050
                )

                # Extract MFCC
                mfcc = librosa.feature.mfcc(
                    y=signal,
                    sr=sample_rate,
                    n_mfcc=40
                )

                # Take mean across time
                mfcc_mean = np.mean(mfcc.T, axis=0)

                features.append(mfcc_mean)
                labels.append(emotion)

            except Exception as e:
                print("Error:", file, e)

print("Feature extraction completed!")

print("Total samples:", len(features))
print("Number of labels:", len(labels))

print("Feature shape:", np.array(features).shape)
```

```python
from collections import Counter

emotion_counts = Counter(labels)

print("Emotion distribution:\n")

for emotion, count in emotion_counts.items():
    print(f"{emotion}: {count}")
```

```python
from sklearn.preprocessing import LabelEncoder

# Convert features and labels to NumPy arrays
X = np.array(features)
y = np.array(labels)

# Encode emotion labels
label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(y)

print("Feature shape:", X.shape)
print("Encoded label shape:", y_encoded.shape)

print("\nEmotion classes:")
for number, emotion in enumerate(label_encoder.classes_):
    print(number, "→", emotion)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y_encoded,
    test_size=0.20,
    random_state=42,
    stratify=y_encoded
)

print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)

print("Training labels:", y_train.shape)
print("Testing labels:", y_test.shape)
```

```python
# Reshape data for CNN
X_train_cnn = X_train[..., np.newaxis]
X_test_cnn = X_test[..., np.newaxis]

print("CNN training shape:", X_train_cnn.shape)
print("CNN testing shape:", X_test_cnn.shape)
```

```python
import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Conv1D, MaxPooling1D
from tensorflow.keras.layers import Flatten, Dense, Dropout

# Create CNN model
model = Sequential([
    Conv1D(
        filters=64,
        kernel_size=3,
        activation="relu",
        input_shape=(40, 1)
    ),

    MaxPooling1D(pool_size=2),

    Conv1D(
        filters=128,
        kernel_size=3,
        activation="relu"
    ),

    MaxPooling1D(pool_size=2),

    Flatten(),

    Dense(128, activation="relu"),

    Dropout(0.3),

    Dense(8, activation="softmax")
])

# Display model architecture
model.summary()
```

```python
import warnings 
warnings.filterwarnings("ignore")
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

print("Model compiled successfully!")
```

```python
history = model.fit(
    X_train_cnn,
    y_train,
    epochs=30,
    batch_size=32,
    validation_split=0.2,
    verbose=1
)
```

```python
test_loss, test_accuracy = model.evaluate(
    X_test_cnn,
    y_test,
    verbose=1
)

print("\nTest Loss:", test_loss)
print("Test Accuracy:", test_accuracy)
print("Test Accuracy (%):", test_accuracy * 100)
```

```python
import os
import numpy as np
import librosa

dataset_path = "audio_speech_actors_01-24"

emotion_map = {
    "01": "neutral",
    "02": "calm",
    "03": "happy",
    "04": "sad",
    "05": "angry",
    "06": "fearful",
    "07": "disgust",
    "08": "surprised"
}

features = []
labels = []

for actor in sorted(os.listdir(dataset_path)):

    actor_path = os.path.join(dataset_path, actor)

    if not os.path.isdir(actor_path):
        continue

    for file in os.listdir(actor_path):

        if file.endswith(".wav"):

            file_path = os.path.join(actor_path, file)

            emotion_code = file.split("-")[2]
            emotion = emotion_map[emotion_code]

            try:
                signal, sample_rate = librosa.load(
                    file_path,
                    sr=22050
                )

                mfcc = librosa.feature.mfcc(
                    y=signal,
                    sr=sample_rate,
                    n_mfcc=40
                )

                mfcc_mean = np.mean(mfcc.T, axis=0)

                features.append(mfcc_mean)
                labels.append(emotion)

            except Exception as e:
                print("Error:", file, e)

print("Feature extraction completed!")
print("Total samples:", len(features))
print("Feature shape:", np.array(features).shape)
```

```python
X = np.array(features)
y = np.array(labels)

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()

y_encoded = label_encoder.fit_transform(y)

print("Encoded labels shape:", y_encoded.shape)

print("\nEmotion classes:")
for number, emotion in enumerate(label_encoder.classes_):
    print(number, "→", emotion)
```

```python
# Reshape data for CNN
X_train_cnn = X_train[..., np.newaxis]
X_test_cnn = X_test[..., np.newaxis]

print("CNN training shape:", X_train_cnn.shape)
print("CNN testing shape:", X_test_cnn.shape)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv1D, MaxPooling1D, Flatten, Dense, Dropout
from tensorflow.keras.optimizers import Adam

# ============================================================
# CNN MODEL
# ============================================================

model = Sequential([
    Conv1D(64, kernel_size=3, activation='relu',
           input_shape=(40, 1)),
    
    MaxPooling1D(pool_size=2),

    Conv1D(128, kernel_size=3, activation='relu'),

    MaxPooling1D(pool_size=2),

    Flatten(),

    Dense(128, activation='relu'),
    Dropout(0.5),

    Dense(8, activation='softmax')
])

# ============================================================
# COMPILE
# ============================================================

model.compile(
    optimizer=Adam(learning_rate=0.001),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# ============================================================
# DISPLAY MODEL
# ============================================================

model.summary()

# ============================================================
# TRAIN CNN
# ============================================================

history = model.fit(
    X_train_cnn,
    y_train,
    epochs=30,
    batch_size=32,
    validation_split=0.2,
    verbose=1
)

# ============================================================
# TEST CNN
# ============================================================

test_loss, test_accuracy = model.evaluate(
    X_test_cnn,
    y_test,
    verbose=0
)

print("\n==============================================")
print("CNN RESULTS")
print("==============================================")
print(f"Test Loss     : {test_loss:.4f}")
print(f"Test Accuracy : {test_accuracy * 100:.2f}%")
```

```python
# ============================================================
# IMPROVED CNN FOR EMOTION RECOGNITION
# ============================================================

import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (
    Conv1D,
    MaxPooling1D,
    BatchNormalization,
    Dropout,
    Dense,
    GlobalAveragePooling1D
)
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import (
    EarlyStopping,
    ReduceLROnPlateau,
    ModelCheckpoint
)
from sklearn.utils.class_weight import compute_class_weight


# ============================================================
# 1. CHECK DATA
# ============================================================

print("Training shape:", X_train_cnn.shape)
print("Testing shape :", X_test_cnn.shape)

print("Training labels:", y_train.shape)
print("Testing labels :", y_test.shape)


# ============================================================
# 2. NORMALIZE MFCC FEATURES
# ============================================================

mean = np.mean(X_train_cnn, axis=0, keepdims=True)
std = np.std(X_train_cnn, axis=0, keepdims=True)

X_train_cnn = (X_train_cnn - mean) / (std + 1e-8)
X_test_cnn = (X_test_cnn - mean) / (std + 1e-8)


# ============================================================
# 3. CALCULATE CLASS WEIGHTS
# ============================================================

classes = np.unique(y_train)

class_weights_array = compute_class_weight(
    class_weight="balanced",
    classes=classes,
    y=y_train
)

class_weights = dict(
    zip(classes, class_weights_array)
)

print("\nClass Weights:")
print(class_weights)


# ============================================================
# 4. BUILD IMPROVED CNN
# ============================================================

model = Sequential([

    # -------------------------------
    # BLOCK 1
    # -------------------------------
    Conv1D(
        filters=64,
        kernel_size=3,
        padding="same",
        activation="relu",
        input_shape=(40, 1)
    ),

    BatchNormalization(),

    Conv1D(
        filters=64,
        kernel_size=3,
        padding="same",
        activation="relu"
    ),

    BatchNormalization(),

    MaxPooling1D(pool_size=2),

    Dropout(0.25),


    # -------------------------------
    # BLOCK 2
    # -------------------------------
    Conv1D(
        filters=128,
        kernel_size=3,
        padding="same",
        activation="relu"
    ),

    BatchNormalization(),

    Conv1D(
        filters=128,
        kernel_size=3,
        padding="same",
        activation="relu"
    ),

    BatchNormalization(),

    MaxPooling1D(pool_size=2),

    Dropout(0.30),


    # -------------------------------
    # BLOCK 3
    # -------------------------------
    Conv1D(
        filters=256,
        kernel_size=3,
        padding="same",
        activation="relu"
    ),

    BatchNormalization(),

    Dropout(0.30),


    # -------------------------------
    # GLOBAL POOLING
    # -------------------------------
    GlobalAveragePooling1D(),


    # -------------------------------
    # FULLY CONNECTED
    # -------------------------------
    Dense(128, activation="relu"),

    BatchNormalization(),

    Dropout(0.40),

    Dense(64, activation="relu"),

    Dropout(0.30),


    # -------------------------------
    # OUTPUT
    # -------------------------------
    Dense(8, activation="softmax")
])


# ============================================================
# 5. COMPILE MODEL
# ============================================================

model.compile(
    optimizer=Adam(learning_rate=0.0005),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)


# ============================================================
# 6. MODEL SUMMARY
# ============================================================

print("\n==============================================")
print("IMPROVED CNN MODEL")
print("==============================================")

model.summary()


# ============================================================
# 7. CALLBACKS
# ============================================================

early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=10,
    restore_best_weights=True,
    verbose=1
)

reduce_lr = ReduceLROnPlateau(
    monitor="val_loss",
    factor=0.5,
    patience=4,
    min_lr=1e-6,
    verbose=1
)

checkpoint = ModelCheckpoint(
    "best_emotion_cnn.keras",
    monitor="val_accuracy",
    save_best_only=True,
    verbose=1
)


# ============================================================
# 8. TRAIN MODEL
# ============================================================

history = model.fit(
    X_train_cnn,
    y_train,

    validation_split=0.20,

    epochs=100,

    batch_size=16,

    class_weight=class_weights,

    callbacks=[
        early_stopping,
        reduce_lr,
        checkpoint
    ],

    verbose=1
)


# ============================================================
# 9. LOAD BEST MODEL
# ============================================================

model = tf.keras.models.load_model(
    "best_emotion_cnn.keras"
)


# ============================================================
# 10. TEST MODEL
# ============================================================

test_loss, test_accuracy = model.evaluate(
    X_test_cnn,
    y_test,
    verbose=0
)


print("\n==============================================")
print("IMPROVED CNN RESULTS")
print("==============================================")

print(f"Test Loss     : {test_loss:.4f}")
print(f"Test Accuracy : {test_accuracy * 100:.2f}%")


# ============================================================
# 11. TRAINING ACCURACY GRAPH
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    history.history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    history.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("Improved CNN Training and Validation Accuracy")

plt.legend()
plt.grid(True)

plt.show()


# ============================================================
# 12. LOSS GRAPH
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Improved CNN Training and Validation Loss")

plt.legend()
plt.grid(True)

plt.show()
```

```python
import numpy as np

# Convert MFCC to a fixed size: 40 MFCCs × 174 time frames
def extract_mfcc_matrix(file_path):
    signal, sample_rate = librosa.load(file_path, sr=22050)

    mfcc = librosa.feature.mfcc(
        y=signal,
        sr=sample_rate,
        n_mfcc=40
    )

    # Fix the number of time frames to 174
    if mfcc.shape[1] < 174:
        mfcc = np.pad(
            mfcc,
            ((0, 0), (0, 174 - mfcc.shape[1])),
            mode="constant"
        )
    else:
        mfcc = mfcc[:, :174]

    return mfcc

# Test on one audio file
test_mfcc = extract_mfcc_matrix(test_file)

print("Fixed MFCC shape:", test_mfcc.shape)
```

```python
import os
import numpy as np
import librosa

features_mfcc = []
labels_mfcc = []

for actor in sorted(os.listdir(dataset_path)):
    actor_path = os.path.join(dataset_path, actor)

    if not os.path.isdir(actor_path):
        continue

    for file in sorted(os.listdir(actor_path)):
        if not file.endswith(".wav"):
            continue

        try:
            # Emotion code from RAVDESS filename
            emotion_code = int(file.split("-")[2])

            # Load audio
            file_path = os.path.join(actor_path, file)
            signal, sample_rate = librosa.load(file_path, sr=22050)

            # Extract MFCC matrix
            mfcc = librosa.feature.mfcc(
                y=signal,
                sr=sample_rate,
                n_mfcc=40
            )

            # Make every sample 40 × 174
            if mfcc.shape[1] < 174:
                mfcc = np.pad(
                    mfcc,
                    ((0, 0), (0, 174 - mfcc.shape[1])),
                    mode="constant"
                )
            else:
                mfcc = mfcc[:, :174]

            features_mfcc.append(mfcc)
            labels_mfcc.append(emotion_code)

        except Exception as e:
            print("Error:", file, e)

features_mfcc = np.array(features_mfcc)
labels_mfcc = np.array(labels_mfcc)

print("Total samples:", len(features_mfcc))
print("Feature shape:", features_mfcc.shape)
print("Labels shape:", labels_mfcc.shape)
```

```python
print("Feature shape:", features_mfcc.shape)
print("Labels shape:", labels_mfcc.shape)

print("\nEmotion class distribution:")

class_counts = Counter(labels_mfcc)

for emotion, count in sorted(class_counts.items()):
    print(f"Class {emotion}: {count} samples")
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    features_mfcc,
    labels_mfcc,
    test_size=0.20,
    random_state=42,
    stratify=labels_mfcc
)

print("Training features shape:", X_train.shape)
print("Testing features shape :", X_test.shape)
print("Training labels shape  :", y_train.shape)
print("Testing labels shape   :", y_test.shape)
```

```python
# Normalize MFCC values
mean = np.mean(X_train)
std = np.std(X_train)

X_train = (X_train - mean) / (std + 1e-8)
X_test = (X_test - mean) / (std + 1e-8)

# Add channel dimension for CNN
X_train = X_train[..., np.newaxis]
X_test = X_test[..., np.newaxis]

print("CNN training shape:", X_train.shape)
print("CNN testing shape :", X_test.shape)

print("Training minimum:", X_train.min())
print("Training maximum:", X_train.max())
```

```python
from tensorflow.keras.utils import to_categorical

# Convert labels from 1–8 to 0–7
y_train_encoded = y_train - 1
y_test_encoded = y_test - 1

# One-hot encode
y_train_cat = to_categorical(y_train_encoded, num_classes=8)
y_test_cat = to_categorical(y_test_encoded, num_classes=8)

print("Encoded training labels shape:", y_train_cat.shape)
print("Encoded testing labels shape :", y_test_cat.shape)

print("\nExample:")
print("Original label:", y_train[0])
print("Encoded label :", y_train_cat[0])
```

```python
from sklearn.utils.class_weight import compute_class_weight
import numpy as np

# Calculate class weights
classes = np.unique(y_train_encoded)

weights = compute_class_weight(
    class_weight="balanced",
    classes=classes,
    y=y_train_encoded
)

class_weights = dict(zip(classes, weights))

print("Class weights:")
for cls, weight in class_weights.items():
    print(f"Class {cls + 1}: {weight:.3f}")
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (
    Conv2D, MaxPooling2D, BatchNormalization,
    Dropout, Flatten, Dense
)

model = Sequential([
    
    # Block 1
    Conv2D(32, (3, 3), activation='relu', padding='same',
           input_shape=(40, 174, 1)),
    BatchNormalization(),
    MaxPooling2D((2, 2)),
    Dropout(0.25),

    # Block 2
    Conv2D(64, (3, 3), activation='relu', padding='same'),
    BatchNormalization(),
    MaxPooling2D((2, 2)),
    Dropout(0.25),

    # Block 3
    Conv2D(128, (3, 3), activation='relu', padding='same'),
    BatchNormalization(),
    MaxPooling2D((2, 2)),
    Dropout(0.30),

    # Classification
    Flatten(),
    Dense(128, activation='relu'),
    BatchNormalization(),
    Dropout(0.40),

    Dense(8, activation='softmax')
])

model.summary()
```

```python
from tensorflow.keras.optimizers import Adam

model.compile(
    optimizer=Adam(learning_rate=0.0005),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

print("Model compiled successfully!")
```

```python
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau

# Stop training if validation accuracy stops improving
early_stopping = EarlyStopping(
    monitor='val_accuracy',
    patience=10,
    mode='max',
    restore_best_weights=True,
    verbose=1
)

# Reduce learning rate when validation loss stops improving
reduce_lr = ReduceLROnPlateau(
    monitor='val_loss',
    factor=0.5,
    patience=4,
    min_lr=1e-6,
    verbose=1
)

# Train the CNN
history = model.fit(
    X_train,
    y_train_cat,
    validation_split=0.15,
    epochs=50,
    batch_size=32,
    class_weight=class_weights,
    callbacks=[early_stopping, reduce_lr],
    verbose=1
)
```

```python
# Evaluate the best restored CNN on the unseen test data
test_loss, test_accuracy = model.evaluate(
    X_test,
    y_test_cat,
    verbose=1
)

print("\n==============================================")
print("FINAL CNN TEST RESULTS")
print("==============================================")
print(f"Test Loss     : {test_loss:.4f}")
print(f"Test Accuracy : {test_accuracy * 100:.2f}%")
```
